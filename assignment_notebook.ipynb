````python
# 시퀀스 데이터 생성 (딥러닝용)
sequence_length = 30

def create_sequences(X, y, seq_len=30):
    X_seq, y_seq = [], []
    for i in range(len(X) - seq_len):
        X_seq.append(X[i:i+seq_len])
        y_seq.append(y[i+seq_len])
    return np.array(X_seq), np.array(y_seq)

X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train.values, sequence_length)
X_val_seq, y_val_seq = create_sequences(X_val_scaled, y_val.values, sequence_length)
X_test_seq, y_test_seq = create_sequences(X_test_scaled, y_test.values, sequence_length)

print(f"시퀀스 데이터 shape: {X_train_seq.shape}")
print(f"  - 샘플 수: {X_train_seq.shape[0]}")
print(f"  - 시퀀스 길이: {X_train_seq.shape[1]}")
print(f"  - 특성 수: {X_train_seq.shape[2]}")
````

## 2. 예제: LSTM 모델 + 확률 기반 포지션 조절 전략

이 예제는 2가지 핵심 요소로 구성됩니다:

1. **예측 모델**: 가격 상승/하락을 예측하고 확률 제공
2. **트레이딩 전략**: 예측 확률에 따라 투자 비율 조절

### 전략 예시
- 상승 확률 90% → 자본의 90% 투자
- 상승 확률 60% → 자본의 60% 투자
- 상승 확률 40% → 투자하지 않음 (현금 보유)

````python
# 예제 모델: LSTM 기반 2-class 분류기
class ExampleLSTMModel(nn.Module):
    """
    예제: LSTM 기반 가격 방향 예측 모델
    출력: 상승 확률 (0~1)
    """
    def __init__(self, input_size, hidden_size=64, dropout=0.2):
        super(ExampleLSTMModel, self).__init__()

        self.lstm1 = nn.LSTM(input_size, hidden_size, batch_first=True, num_layers=1)
        self.dropout1 = nn.Dropout(dropout)
        self.bn1 = nn.BatchNorm1d(hidden_size)

        self.lstm2 = nn.LSTM(hidden_size, hidden_size//2, batch_first=True, num_layers=1)
        self.dropout2 = nn.Dropout(dropout)
        self.bn2 = nn.BatchNorm1d(hidden_size//2)

        self.fc1 = nn.Linear(hidden_size//2, 16)
        self.relu = nn.ReLU()
        self.dropout3 = nn.Dropout(dropout)

        self.fc2 = nn.Linear(16, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        lstm_out, _ = self.lstm1(x)
        lstm_out = self.dropout1(lstm_out)
        lstm_out = lstm_out.permute(0, 2, 1)
        lstm_out = self.bn1(lstm_out)
        lstm_out = lstm_out.permute(0, 2, 1)

        lstm_out, _ = self.lstm2(lstm_out)
        lstm_out = self.dropout2(lstm_out[:, -1, :])
        lstm_out = self.bn2(lstm_out)

        out = self.fc1(lstm_out)
        out = self.relu(out)
        out = self.dropout3(out)

        out = self.fc2(out)
        out = self.sigmoid(out)

        return out

# 예제 모델 생성
example_model = ExampleLSTMModel(
    input_size=X_train_seq.shape[2],
    hidden_size=64,
    dropout=0.2
).to(device)

print("예제 모델 구조:")
print(example_model)
print(f"\nTotal parameters: {sum(p.numel() for p in example_model.parameters()):,}")
````

````python
# 학습 함수 (제공됨)
def train_model(model, train_loader, val_loader, epochs=100, lr=0.001, patience=15):
    model = model.to(device)
    criterion = nn.BCELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': []}
    best_val_loss = float('inf')
    patience_counter = 0
    best_model_state = None

    for epoch in range(epochs):
        model.train()
        train_loss = 0
        train_correct = 0
        train_total = 0

        for batch_X, batch_y in train_loader:
            batch_X, batch_y = batch_X.to(device), batch_y.to(device)
            optimizer.zero_grad()
            outputs = model(batch_X)
            loss = criterion(outputs, batch_y.unsqueeze(1))
            loss.backward()
            optimizer.step()

            train_loss += loss.item()
            predicted = (outputs > 0.5).float()
            train_total += batch_y.size(0)
            train_correct += (predicted.squeeze() == batch_y).sum().item()

        avg_train_loss = train_loss / len(train_loader)
        train_acc = train_correct / train_total

        model.eval()
        val_loss = 0
        val_correct = 0
        val_total = 0

        with torch.no_grad():
            for batch_X, batch_y in val_loader:
                batch_X, batch_y = batch_X.to(device), batch_y.to(device)
                outputs = model(batch_X)
                loss = criterion(outputs, batch_y.unsqueeze(1))

                val_loss += loss.item()
                predicted = (outputs > 0.5).float()
                val_total += batch_y.size(0)
                val_correct += (predicted.squeeze() == batch_y).sum().item()

        avg_val_loss = val_loss / len(val_loader)
        val_acc = val_correct / val_total

        history['train_loss'].append(avg_train_loss)
        history['val_loss'].append(avg_val_loss)
        history['train_acc'].append(train_acc)
        history['val_acc'].append(val_acc)

        if (epoch + 1) % 10 == 0:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}, Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f}')

        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            patience_counter = 0
            best_model_state = model.state_dict().copy()
        else:
            patience_counter += 1

        if patience_counter >= patience:
            print(f'Early stopping at epoch {epoch+1}')
            break

    if best_model_state is not None:
        model.load_state_dict(best_model_state)

    return history

print("✅ 학습 함수 준비 완료!")
````

````python
# 예제 모델 학습
print("예제 모델 학습 시작...")
example_history = train_model(
    model=example_model,
    train_loader=train_loader,
    val_loader=val_loader,
    epochs=100,
    lr=0.001,
    patience=15
)
print("\n✅ 예제 모델 학습 완료!")
````

````python
# 학습 과정 시각화
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].plot(example_history["train_loss"], label="Train Loss", linewidth=2)
axes[0].plot(example_history["val_loss"], label="Validation Loss", linewidth=2)
axes[0].set_title("Example Model Loss", fontsize=14, fontweight="bold")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(example_history["train_acc"], label="Train Accuracy", linewidth=2)
axes[1].plot(example_history["val_acc"], label="Validation Accuracy", linewidth=2)
axes[1].set_title("Example Model Accuracy", fontsize=14, fontweight="bold")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Accuracy")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## 3. 트레이딩 전략: 확률 기반 포지션 조절

예측 확률에 따라 투자 비율을 조절하는 전략입니다.

````python
# 예측 함수 (확률 포함)
def predict_with_probability(model, data_loader):
    """
    모델 예측 및 확률 반환
    """
    model.eval()
    predictions_prob = []

    with torch.no_grad():
        for batch_X, _ in data_loader:
            batch_X = batch_X.to(device)
            outputs = model(batch_X)
            predictions_prob.append(outputs.cpu().numpy())

    predictions_prob = np.vstack(predictions_prob).flatten()
    predictions = (predictions_prob > 0.5).astype(int)

    return predictions_prob, predictions

# 예제 모델 예측
example_prob, example_pred = predict_with_probability(example_model, test_loader)

print(f"예측 완료!")
print(f"예측 수: {len(example_pred)}")
print(f"상승 예측: {np.sum(example_pred == 1)}개")
print(f"하락 예측: {np.sum(example_pred == 0)}개")
print(f"\n평균 상승 확률: {example_prob.mean():.2%}")
````

````python
# 확률 기반 트레이딩 전략
def simulate_probability_trading(predictions_prob, actual_prices, dates,
                                 initial_capital=10000, transaction_fee=0.001,
                                 threshold=0.5, position_scaling=True):
    """
    확률 기반 트레이딩 시뮬레이션

    Parameters:
    -----------
    predictions_prob : array
        상승 확률 (0~1)
    actual_prices : array
        실제 가격
    dates : array
        날짜
    initial_capital : float
        초기 자본
    transaction_fee : float
        거래 수수료
    threshold : float
        매수 결정 임계값 (기본 0.5)
    position_scaling : bool
        True: 확률에 비례하여 투자 비율 조절
        False: 임계값 초과 시 전액 투자
    """
    cash = initial_capital
    btc_holdings = 0
    portfolio_values = []
    trade_log = []

    for i in range(len(predictions_prob)):
        current_price = actual_prices[i]
        prob = predictions_prob[i]

        portfolio_value = cash + btc_holdings * current_price
        portfolio_values.append(portfolio_value)

        # 마지막 날 전량 매도
        if i == len(predictions_prob) - 1:
            if btc_holdings > 0:
                sell_value = btc_holdings * current_price * (1 - transaction_fee)
                trade_log.append({
                    'date': dates[i],
                    'action': 'SELL_ALL',
                    'price': current_price,
                    'prob': prob,
                    'amount': btc_holdings,
                    'value': btc_holdings * current_price,
                    'fee': btc_holdings * current_price * transaction_fee
                })
                cash += sell_value
                btc_holdings = 0
            continue

        # 투자 비율 결정
        if position_scaling:
            # 확률 기반 포지션 조절
            if prob > threshold:
                # 확률에 비례하여 투자 (예: 70% 확률 → 70% 투자)
                invest_ratio = prob
            else:
                # 임계값 미만이면 매도 또는 관망
                invest_ratio = 0
        else:
            # 단순 전략: 임계값 초과 시 전액 투자
            invest_ratio = 1.0 if prob > threshold else 0.0

        # 현재 포지션 비율
        current_position_ratio = (btc_holdings * current_price) / portfolio_value if portfolio_value > 0 else 0

        # 포지션 조정
        target_btc_value = portfolio_value * invest_ratio
        current_btc_value = btc_holdings * current_price

        if target_btc_value > current_btc_value:  # 매수 필요
            buy_cash = target_btc_value - current_btc_value
            if buy_cash > cash:
                buy_cash = cash
            if buy_cash > 0:
                buy_amount = (buy_cash * (1 - transaction_fee)) / current_price
                btc_holdings += buy_amount
                trade_log.append({
                    'date': dates[i],
                    'action': 'BUY',
                    'price': current_price,
                    'prob': prob,
                    'amount': buy_amount,
                    'value': buy_cash,
                    'fee': buy_cash * transaction_fee
                })
                cash -= buy_cash
        elif target_btc_value < current_btc_value:  # 매도 필요
            sell_btc = (current_btc_value - target_btc_value) / current_price
            if sell_btc > btc_holdings:
                sell_btc = btc_holdings
            if sell_btc > 0:
                sell_value = sell_btc * current_price * (1 - transaction_fee)
                trade_log.append({
                    'date': dates[i],
                    'action': 'SELL',
                    'price': current_price,
                    'prob': prob,
                    'amount': sell_btc,
                    'value': sell_btc * current_price,
                    'fee': sell_btc * current_price * transaction_fee
                })
                cash += sell_value
                btc_holdings -= sell_btc

    final_value = portfolio_values[-1]
    total_return = (final_value - initial_capital) / initial_capital * 100

    total_trade_volume = sum(trade['value'] for trade in trade_log)
    total_fees_paid = sum(trade['fee'] for trade in trade_log)

    return {
        'initial_capital': initial_capital,
        'final_value': final_value,
        'total_return': total_return,
        'portfolio_values': portfolio_values,
        'trade_log': trade_log,
        'num_trades': len(trade_log),
        'total_trade_volume': total_trade_volume,
        'total_fees_paid': total_fees_paid
    }

print("✅ 트레이딩 시뮬레이션 함수 준비 완료!")
````

```
1. 모델 아키텍처:
   - 본 과제에서 설계한 MyTradingModel은 양방향 LSTM(BiLSTM)과 Attention 메커니즘을 결합한 시계열 분류 모델이다. 입력은 (batch_size, sequence_length, feature_dim) 형태의 가격/기술지표 시퀀스이고, 출력은 다음 시점 가격이 상승할 확률을 나타내는 스칼라 값(0~1)이다. 구성은 다음과 같다.

   1-1. 입력층
    - input_size = feature_dim
    - 여러 가지 가격 및 기술 지표(feature)를 그대로 입력받는다.

   1-2. BiLSTM 인코더
    - nn.LSTM(input_size, hidden_size, numlayers=num_layers, bidirectional=True, batch_first=True)
    - 정방향/역방향 정보를 동시에 학습하여 각 타임스텝마다 과거와 상대적 미래의 문맥을 함께 반영한다.
    - 출력 크기는 (batch_size, sequence_length, 2 * hidden_size)이다.

   1-3. Attention 레이어
    - 각 타임스텝의 은닉 상태 h_t ∈ R^{2H}에 대해 score_t = Linear(2H -> 1)을 통과시켜 중요도를 계산한 후, 시퀀스 방향으로 softmax를 적용하여 attention weight을 구한다.
    - 이 가중치를 BiLSTM 출력에 곱하여, 전체 시퀀스를 하나의 context vector(가중합)로 요약한다. 결과 크기는 (batch_size, 2 * hidden_size)

   1-4. Fully-connected 및 출력층
    - Linear(2H -> 32) -> ReLU -> Dropout
    - Linear(32 -> 1) -> Sigmoid
    - 최종적으로 0~1 사이의 실수 값을 출력하며, 이를 다음 시점 가격 상승 확률로 해석한다.

2. 선택 이유:
   - 양방향 LSTM 사용 이유 : 단순히 마지막 타임스텝의 은닉 상태만 사용하는 것보다 전체 시퀀스의 패턴을 양방향으로 보는 것이 추세 + 반전 패턴을 파악하는데 유리하다고 판단하였다. 특히 가격 데이터는 단기 노이즈가 많기 때문에 과거 전체 구간 패턴을 종합적으로 요약해주는 BiLSTM이 적합하다고 보았다.

   - Attention 도입 이유 : 모든 타임스텝이 동일하게 중요하지는 않다. 직전 급등/급락 구간, 거래량이 크게 튀는 시점 등은 미래 방향성 예측에 훨씬 중요한 정보가 된다. 이 때 Attention을 통해 어떤 구간에 더 주목해야 하는지를 모델이 스스로 학습하도록 하여 단순히 마지막 타임스텝만 보는 모델보다 더 많은 정보를 활용할 수 있게 했다.

   - 얕은 MLP 헤드(32차원 -> 1)을 선택한 이유 : 출력단은 과도하게 깊게 설계하지 않고, BiLSTM+Attention에서 이미 추출된 시계열 표현을 가볍게 비선형 변환만 하도록 설계했다. 이는 파라미터 수를 필요 이상으로 늘리지 않으면서도 비선형 의사결정 경계를 학습하기에 충분하다고 판단하였다.

3. 트레이딩 전략:
   - 모델이 예측한 상승 확률 p = model(x)를 기반으로, 다음과 같은 간단하지만 리스크를 고려한 트레이딩 전략을 사용했다.

   3-1. 진입/관망 기준 (threshold 전략)
    - 예측 확률 p가 일정 기준값 threshold를 초과할 때만 매수 포지션을 진입한다.
    - p > threshold -> 매수(long)
    - p ≤ threshold -> 포지션 진입/유지 안 함(관망 또는 기존 포지션 청산)
    - threshold를 0.5보다 약간 높게 설정하여 모델이 상승일 가능성이 높다고 강하게 확신하는 경우에만 거래하도록 만들어 리스크 감소

   3-2. Position scaling(포지션 크기 조절)
    - 단순히 매수/관망만 나누는 것이 아닌 p가 0.5에서 얼마나 떨어져 있는지에 따라 포지션 크기를 조절한다.
    - 예시: position_size = max(0, p - 0.5) / (1 - 0.5) -> p = 0.5일 때 0, p = 1일 때 1이 되도록 스케일링
    - 이렇게 하면 확신이 강할수록 더 큰 비중으로 진입하고, 애매한 구간에서는 자연스럽게 포지션이 줄어들어 리스크가 완화된다.

   3-3. 수수료 및 연속 거래 고려
    - 제공된 시뮬레이션 함수에서 거래 수수료를 명시적으로 반영하여 지나치게 잦은 매매가 오히려 수익을 갉아먹지 않도록 했다.
    - threshold와 position scaling을 함께 사용하면 변동성이 크더라도 불필요한 잡거래를 줄이는 효과가 있다.

4. 하이퍼파라미터:
   - hidden_size: 64
   - learning_rate: 0.001
   - threshold: 0.55
   - position_scaling: True

5. 예제와의 차별점:
   - 예제와의 차별점은 다음과 같다.

   5-1. 양방향 vs 단방향
    - 예제 모델은 단방향 LSTM 레이어를 여러 층 쌓은 구조이지만 제안한 모델은 BiLSTM을 사용하여 각 시점이 과거뿐 아니라 시퀀스 전체 문맥을 반영하도록 했다.

   5-2. 마지막 타임스텝 vs Attention 기반 요약
    - 예제 모델은 대체로 마지막 타임스텝의 은닉 상태만 사용해서 시퀀스를 요약한다. MyTradingModel은 Attention을 통해 급변 구간, 특정 패턴(급락 후 반등 등)이 나타나는 구간에 더 큰 가중치를 주어 전체 시퀀스에서 중요한 부분만 강조된 context vector를 사용한다.

   5-3. 트레이딩 관점에서의 활용
    - 예제 모델은 확률 예측에 초점을 두고 있고, 트레이딩 전략은 상대적으로 단순한 threshold 기반 매수/관망에 머무른다. MyTradingModel은 position_scaling을 기본 전제로 설계했으며, 예측 확률의 크기를 그대로 포지션 크기와 연동하여 리스크와 레버리지 개념을 함께 반영하려 했다. 즉, 모델 구조뿐 아니라 출력 확률을 어떻게 트레이딩 전략에 연결할지까지 고려한 설계라는 점이 차별점이다.

   5-4. 파라미터 효율 및 일반화
    - 예제 모델은 LSTM 2층 + BatchNorm + FC 2층 등 비교적 깊은 구조이다. MyTradingModel은 BiLSTM + Attention + 얕은 MLP 형태로, 시퀀스 표현부에 집중하고 출력단은 간결하게 유지하여 과도한 파라미터 증가 없이 일반화 성능을 노린 구조이다.
```

````python
# TODO: 자신만의 모델 구현

class MyTradingModel(nn.Module):
    """
    TODO: 자신만의 트레이딩 모델 구현

    입력: (batch, sequence_length, features)
    출력: (batch, 1) - 상승 확률 (0~1)
    """
    def __init__(self, input_size, hidden_size=64, dropout=0.3, num_layers=2):
        super(MyTradingModel, self).__init__()

        # 양방향 LSTM: 과거 전체 시퀀스 정보를 양쪽 방향으로 학습
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            bidirectional=True
        )

        # Attention: 각 타임스텝의 중요도를 학습
        self.attn = nn.Linear(hidden_size * 2, 1)

        # Fully-connected layers
        self.dropout = nn.Dropout(dropout)
        self.fc1 = nn.Linear(hidden_size * 2, 32)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(32, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # TODO: Forward pass 구현
        # x: (batch, seq_len, features)
        lstm_out, _ = self.lstm(x)  # (B, T, 2H) (bidirectional)

        # Attention score 계산
        # 각 타임스텝마다 중요도 점수 -> softmax로 정규화
        attn_scores = self.attn(lstm_out) # (B, T, 1)
        attn_weights = torch.softmax(attn_scores, dim=1)  # (B, T, 1)

        # 가중합 -> context vector
        context = torch.sum(attn_weights * lstm_out, dim=1) # (B, 2H)

        # 완전연결 신경망 통과
        out = self.dropout(context)
        out = self.fc1(out)
        out = self.relu(out)
        out = self.dropout(out)
        out = self.fc2(out) # (B, 1)
        out = self.sigmoid(out) # 0~1 확률

        return out

# TODO: 모델 생성 (주석 해제 후 사용)
my_model = MyTradingModel(
     input_size=X_train_seq.shape[2],
     hidden_size=64,
     dropout=0.2
).to(device)

print("나의 모델 구조:")
print(my_model)
print(f"\nTotal parameters: {sum(p.numel() for p in my_model.parameters()):,}")
````

**3. 모델 설계**

```
- 아키텍처 선택 이유: BiLSTM + Attention을 모델에 적용하였다. BiLSTM은 단방향 LSTM 대신 양방향 LSTM을 사용하여 각 시점이 시퀀스의 앞/뒤 문맥을 모두 반영하도록 했다. 비트코인 가격처럼 노이즈가 많은 시계열에서는 특정 구간의 패턴을 앞뒤 맥락으로보고 판단하는 것이 더 유리하다고 판단했다. Attention은 모든 타임스텝이 동일하게 중요한 것은 아니므로, 모델이 어떤 구간을 더 중요하게 볼지를 스스로 학습하도록 단순한 점수 기반 Attention 레이어를 추가했다. 이로 인해 급격한 가격 변화나 추세 전환이 나타나는 구간이 context vector에 더 크게 반영될 수 있다. 또한 얕은 MLP 출력층을 사용하여 파라미터 수를 과도하게 늘리지 않고 일반화 성능을 확보하고자 했다.

- 하이퍼파라미터 튜닝:
  hidden_size = 64
    - 너무 작으면 복잡한 가격 패턴을 표현하기 어렵고, 너무 크면 과적합 및 계산량 증가 문제가 생기므로 중간 수준인 64를 선택했다. BiLSTM이므로 실제로는 2x64 차원의 표현을 사용한다.

  learning_rate = 0.001
    - Adam 옵티마이저에서 안정적으로 자주 사용되는 기본 값을 채택하였다.

  threshold ≈ 0.55
    - 0.5보다 약간 높게 설정하여 모델이 확실히 상승 쪽이 더 유리하다고 판단하는 경우에만 진입하도록 했다.

  position_scaling = True
    - 확률 값과 포지션 크기를 직접 연결하여 확신도가 높을수록 더 큰 비중으로 진입할 수 있게 했다. 이를 통해 예측 확률을 그냥 이진 분류 결과로만 쓰지 않고 리스크/레버리지 조절에도 활용하는 방향으로 설계했다.

- 예제 모델과의 차이점: 예제 모델은 단방향 LSTM 2층 + BatchNorm + FC이고 MyTradingModel은 BiLSTM + Attention + 얕은 FC이다. 구체적인 차이로 예제 모델은 단방향, MyTradingModel은 양방향이며 시퀀스 요약 방식이 예제 모델은 주로 마지막 타임스텝을 사용하지만 MyTradingModel은 Attention을 통해 전체 타임스텝 중 중요한 구간에 더 큰 가중치를 부여한다. 또한 예제 모델은 확률 출력을 단순 threshold로 이진화하는데 초점을 두었지만 MyTradingModel은 확률 값을 그대로 포지션 스케일링에 활용하도록 설계하여 모델 출력과 전략이 더 밀접하게 연결괸다.
```
